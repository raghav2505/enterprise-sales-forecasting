```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
print("Libraries Imported Successfully")
```

```python
sales_df = train.copy()
sales_df["year"] = sales_df["date"].dt.year
sales_df["month"] = sales_df["date"].dt.month
sales_df["day"] = sales_df["date"].dt.day
sales_df["day_of_week"] = sales_df["date"].dt.dayofweek

sales_df[
    ["date", "year", "month", "day", "day_of_week"]
].head()
```

```python
sales_df[
    ["date", "store_nbr", "family", "sales", "onpromotion"]
].head(10)
```

```python
print("Stores:", sales_df["store_nbr"].nunique())
print("Product families:", sales_df["family"].nunique())
print("Unique dates:", sales_df["date"].nunique())
```

```python
model_df = sales_df[
    [
        "date",
        "store_nbr",
        "family",
        "sales",
        "onpromotion",
        "year",
        "month",
        "day",
        "day_of_week"
    ]
].copy()

print("Model dataset shape:", model_df.shape)
print("\nColumns:")
print(model_df.columns.tolist())

model_df.head()
```

```python
model_df = model_df.sort_values("date")

split_date = model_df["date"].quantile(0.85)

train_df = model_df[model_df["date"] < split_date].copy()
valid_df = model_df[model_df["date"] >= split_date].copy()

print("Split date:", split_date)
print("Training rows:", len(train_df))
print("Validation rows:", len(valid_df))
print("Training ends:", train_df["date"].max())
print("Validation starts:", valid_df["date"].min())
```

```python
print("Training categories:", train_df["family"].nunique())
print("Validation categories:", valid_df["family"].nunique())

unseen_families = (
    set(valid_df["family"].unique())
    - set(train_df["family"].unique())
)

print("Unseen validation categories:", unseen_families)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

feature_cols = [
    "store_nbr",
    "family",
    "onpromotion",
    "year",
    "month",
    "day",
    "day_of_week"
]

X_train = train_df[feature_cols]
y_train = train_df["sales"]

X_valid = valid_df[feature_cols]
y_valid = valid_df["sales"]

X_train = pd.get_dummies(X_train, columns=["family"])
X_valid = pd.get_dummies(X_valid, columns=["family"])

X_valid = X_valid.reindex(columns=X_train.columns, fill_value=0)

print("Training features:", X_train.shape)
print("Validation features:", X_valid.shape)
print("Features prepared successfully!")
```

```python
dt_model = DecisionTreeRegressor(
    max_depth=12,
    random_state=42
)

dt_model.fit(X_train, y_train)

dt_predictions = dt_model.predict(X_valid)

mae = mean_absolute_error(y_valid, dt_predictions)
rmse = np.sqrt(mean_squared_error(y_valid, dt_predictions))

print("Decision Tree Results")
print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_valid)

rf_mae = mean_absolute_error(y_valid, rf_predictions)
rf_rmse = np.sqrt(mean_squared_error(y_valid, rf_predictions))

print("Random Forest Results")
print("MAE:", round(rf_mae, 2))
print("RMSE:", round(rf_rmse, 2))
```

```python
X_train_small = X_train.sample(n=20000, random_state=42)
y_train_small = y_train.loc[X_train_small.index]

X_valid_small = X_valid.sample(n=10000, random_state=42)
y_valid_small = y_valid.loc[X_valid_small.index]

rf_model = RandomForestRegressor(
    n_estimators=10,
    max_depth=8,
    min_samples_leaf=10,
    random_state=42,
    n_jobs=2
)

rf_model.fit(X_train_small, y_train_small)

rf_predictions = rf_model.predict(X_valid_small)

print("MAE:", round(mean_absolute_error(y_valid_small, rf_predictions), 2))
print("RMSE:", round(np.sqrt(mean_squared_error(y_valid_small, rf_predictions)), 2))
```

```python
sales_df = sales_df.sort_values(["store_nbr", "family", "date"]).copy()

sales_df["sales_lag_7"] = sales_df.groupby(
    ["store_nbr", "family"]
)["sales"].shift(7)

sales_df["sales_lag_14"] = sales_df.groupby(
    ["store_nbr", "family"]
)["sales"].shift(14)

sales_df[[
    "date", "store_nbr", "family", "sales",
    "sales_lag_7", "sales_lag_14"
]].head(20)
```

```python
print(sales_df[["sales_lag_7", "sales_lag_14"]].isnull().sum())

print("Original rows:", len(sales_df))

print("Rows with both lag values:", len(sales_df.dropna(subset=["sales_lag_7", "sales_lag_14"])))
```

```python
sales_df = sales_df.dropna(subset=["sales_lag_7", "sales_lag_14"]).copy()

print("Clean dataset shape:", sales_df.shape)
print("Missing lag values:", sales_df[["sales_lag_7", "sales_lag_14"]].isnull().sum().sum())
```

```python
sales_df = sales_df.sort_values("date").copy()

split_date = sales_df["date"].quantile(0.85)

train_df = sales_df[sales_df["date"] < split_date].copy()
valid_df = sales_df[sales_df["date"] >= split_date].copy()

print("Split date:", split_date)
print("Training rows:", len(train_df))
print("Validation rows:", len(valid_df))
print("Training ends:", train_df["date"].max())
print("Validation starts:", valid_df["date"].min())
```

```python
feature_cols = [
    "store_nbr",
    "family",
    "onpromotion",
    "year",
    "month",
    "day",
    "day_of_week",
    "sales_lag_7",
    "sales_lag_14"
]

X_train = train_df[feature_cols].copy()
y_train = train_df["sales"].copy()

X_valid = valid_df[feature_cols].copy()
y_valid = valid_df["sales"].copy()

X_train = pd.get_dummies(X_train, columns=["family"], dtype=int)
X_valid = pd.get_dummies(X_valid, columns=["family"], dtype=int)

X_valid = X_valid.reindex(columns=X_train.columns, fill_value=0)

print("Training features:", X_train.shape)
print("Validation features:", X_valid.shape)
print("Training target:", y_train.shape)
print("Validation target:", y_valid.shape)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

X_train_small = X_train.sample(n=20000, random_state=42)
y_train_small = y_train.loc[X_train_small.index]

X_valid_small = X_valid.sample(n=10000, random_state=42)
y_valid_small = y_valid.loc[X_valid_small.index]

dt_model = DecisionTreeRegressor(
    max_depth=12,
    random_state=42
)

dt_model.fit(X_train_small, y_train_small)

dt_predictions = dt_model.predict(X_valid_small)

dt_mae = mean_absolute_error(y_valid_small, dt_predictions)
dt_rmse = np.sqrt(mean_squared_error(y_valid_small, dt_predictions))

print("Decision Tree Results")
print("MAE:", round(dt_mae, 2))
print("RMSE:", round(dt_rmse, 2))
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=10,
    max_depth=10,
    min_samples_leaf=10,
    random_state=42,
    n_jobs=2
)

rf_model.fit(X_train_small, y_train_small)

rf_predictions = rf_model.predict(X_valid_small)

rf_mae = mean_absolute_error(y_valid_small, rf_predictions)
rf_rmse = np.sqrt(mean_squared_error(y_valid_small, rf_predictions))

print("Random Forest Results")
print("MAE:", round(rf_mae, 2))
print("RMSE:", round(rf_rmse, 2))
```

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=100,
    max_depth=8,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=2
)

xgb_model.fit(X_train_small, y_train_small)

xgb_predictions = xgb_model.predict(X_valid_small)

xgb_mae = mean_absolute_error(y_valid_small, xgb_predictions)
xgb_rmse = np.sqrt(mean_squared_error(y_valid_small, xgb_predictions))

print("XGBoost Results")
print("MAE:", round(xgb_mae, 2))
print("RMSE:", round(xgb_rmse, 2))
```

```python
results = pd.DataFrame({
    "Model": ["Decision Tree", "Random Forest", "XGBoost"],
    "MAE": [dt_mae, rf_mae, xgb_mae],
    "RMSE": [dt_rmse, rf_rmse, xgb_rmse]
})

results = results.sort_values("RMSE").reset_index(drop=True)

print(results.round(2).to_string(index=False))

plt.figure(figsize=(9, 5))
plt.bar(results["Model"], results["RMSE"])
plt.title("Model Comparison - RMSE")
plt.xlabel("Model")
plt.ylabel("RMSE")
plt.tight_layout()
plt.show()
```
